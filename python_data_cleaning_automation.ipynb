```python
import pandas as pd

data = {
    "Name": ["Juan", "Maria", "Pedro", "Ana", "Maria"],
    "Age": [22, 23, None, 24, 23],
    "Department": ["IT", "it", "IT", "HR", "it"],
    "Score": [85, 90, 85, 88, 90]
}

df = pd.DataFrame(data)

print(df)
```

```python
# 1. Remove duplicate records
df = df.drop_duplicates()

# 2. Fill missing ages with the average age
df["Age"] = df["Age"].fillna(df["Age"].mean())

# 3. Standardize department names
df["Department"] = df["Department"].str.upper()

print("CLEANED DATA:")
print(df)
```

```python
# Calculate basic statistics

average_score = df["Score"].mean()
average_age = df["Age"].mean()

department_counts = df["Department"].value_counts()

print("DATA SUMMARY")
print("--------------------")
print(f"Average Age: {average_age:.1f}")
print(f"Average Score: {average_score:.1f}")

print("\nNumber of records per department:")
print(department_counts)
```

```python
# Create an automatic report

print("\n===== AUTOMATED DATA REPORT =====")

print(f"Total records: {len(df)}")
print(f"Average age: {df['Age'].mean():.1f}")
print(f"Average score: {df['Score'].mean():.1f}")

print("\nDepartment breakdown:")
for department, count in df["Department"].value_counts().items():
    print(f"- {department}: {count} record(s)")

print("\nData cleaning completed successfully!")
```

```python
# Save the cleaned data to a new CSV file

df.to_csv("cleaned_data.csv", index=False)

print("Cleaned data has been saved as cleaned_data.csv")
```

```python
# Check the quality of the cleaned data

print("===== DATA QUALITY CHECK =====")

missing_values = df.isnull().sum().sum()
duplicate_rows = df.duplicated().sum()

print(f"Missing values: {missing_values}")
print(f"Duplicate rows: {duplicate_rows}")

if missing_values == 0 and duplicate_rows == 0:
    print("Data quality check: PASSED")
else:
    print("Data quality check: NEEDS REVIEW")
```
